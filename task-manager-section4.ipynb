```python
# Section 4.1 - Menu Display Function

def display_menu():
    """
    Display task options to the user.

    >>> display_menu()  # doctest: +SKIP
    1. Add task
    2. List tasks
    3. Remove task
    4. Quit
    >>> display_menu()  # doctest: +SKIP
    """
    print("1. Add task")
    print("2. List tasks")
    print("3. Remove task")
    print("4. Quit")

```

```python
# Section 4.1 - User Choice Input


def get_user_choice():
    """
    Prompt user for menu choice with validation.

    >>> get_user_choice()  # doctest: +SKIP
    2
    >>> get_user_choice()  # doctest: +SKIP
    4
    """
    while True:
        choice = input("Enter your choice (1-4): ")
        if choice.isdigit():
            choice = int(choice)
            if 1 <= choice <= 4:
                return choice
            else:
                print("Please enter a number between 1 and 4.")
        else:
            print("Invalid input. Please enter a number.")

```

```python
# Section 4.1 - Task Manager Main Loop

def main():
    """
    Main program loop for task manager.
    """
    tasks = []
    running = True

    while running:
        print("\n--- MENU ---")
        display_menu()
        choice = get_user_choice()

        if choice == 1:
            task = input("Enter a new task: ")
            tasks.append(task)
            print(f"Added: {task}")

        elif choice == 2:
            print("\n--- YOUR TASKS ---")
            if len(tasks) == 0:
                print("No tasks to show.")
            else:
                for i in range(len(tasks)):
                    print(f"{i}. {tasks[i]}")

        elif choice == 3:
            print("\n--- REMOVE TASK ---")
            if len(tasks) == 0:
                print("No tasks to remove.")
            else:
                index_input = input("Enter the task index to remove: ")
                if index_input.isdigit():
                    index = int(index_input)
                    if 0 <= index < len(tasks):
                        removed = tasks.pop(index)
                        print(f"Removed: {removed}")
                    else:
                        print("Invalid index.")
                else:
                    print("Please enter a number.")

        elif choice == 4:
            print("\nGoodbye!")
            running = False


# Uncomment to run interactively
main()
```
