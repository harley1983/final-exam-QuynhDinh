# Section 1 - Iterative Prompt Engineering  

**Initial Prompt:**  I need to make a basic CLI program in Python that lets a user manage tasks. The user should be able to add new tasks, view all tasks, and delete a task. The tasks are stored in memory.
Please give me beginner-friendly pseudocode using only basic Python features like lists, dictionaries, functions, and input/output.


````python
# Task Manager CLI - Pseudocode

## Data Structure
```
tasks = []  # List to store all tasks in memory
task_counter = 1  # To assign unique IDs to tasks
```

## Main Program Flow
```
WHILE program is running:
    DISPLAY menu options:
        1. Add Task
        2. View All Tasks
        3. Delete Task
        4. Exit

    GET user choice

    IF choice is 1:
        CALL add_task function
    ELSE IF choice is 2:
        CALL view_tasks function
    ELSE IF choice is 3:
        CALL delete_task function
    ELSE IF choice is 4:
        PRINT "Goodbye!"
        EXIT program
    ELSE:
        PRINT "Invalid choice, please try again"
```

## Function: add_task()
```
FUNCTION add_task():
    PROMPT user for task description
    GET task_description from user input

    IF task_description is not empty:
        CREATE new_task dictionary:
            id = task_counter
            description = task_description

        ADD new_task to tasks list
        INCREMENT task_counter by 1
        PRINT "Task added successfully!"
    ELSE:
        PRINT "Task description cannot be empty"
```

## Function: view_tasks()
```
FUNCTION view_tasks():
    IF tasks list is empty:
        PRINT "No tasks available"
        RETURN

    PRINT "Your Tasks:"
    FOR each task in tasks:
        PRINT task id and task description
        FORMAT: "ID: [id] - [description]"
```

## Function: delete_task()
```
FUNCTION delete_task():
    IF tasks list is empty:
        PRINT "No tasks to delete"
        RETURN

    CALL view_tasks to show current tasks

    PROMPT user for task ID to delete
    GET task_id from user input

    SET found = False
    FOR each task in tasks:
        IF task id equals task_id:
            REMOVE task from tasks list
            SET found = True
            PRINT "Task deleted successfully!"
            BREAK out of loop

    IF found is False:
        PRINT "Task with that ID not found"
```

## Helper Function: get_user_input()
```
FUNCTION get_user_input(prompt):
    TRY:
        GET input from user with prompt
        RETURN input
    CATCH any errors:
        RETURN empty string
```

## Error Handling Notes
- Always check if user input is valid
- Handle cases where task list is empty
- Validate task IDs when deleting
- Use try-catch for input operations to prevent crashes

## Sample Task Dictionary Structure
```
task = {
    "id": 1,
    "description": "Buy groceries"
}
```
````

# Section 1.2 - First Refinement Prompt  


**Refinement Prompt 1:**  
- Please improve the pseudocode by applying better input handling and menu logic. Specifically:

- Use a function to validate that user choices are integers and within valid options (1–4).

- Prevent the program from crashing on invalid input by using try/except.

- Keep the main menu loop running unless the user explicitly chooses Exit.



````python
# Task Manager CLI - Pseudocode

## Data Structure
```
tasks = []  # List to store all tasks in memory
task_counter = 1  # To assign unique IDs to tasks
```

## Main Program Flow
```
SET running = True

WHILE running is True:
    DISPLAY menu options:
        1. Add Task
        2. View All Tasks
        3. Delete Task
        4. Exit

    GET valid_choice from get_menu_choice function

    IF valid_choice is 1:
        CALL add_task function
    ELSE IF valid_choice is 2:
        CALL view_tasks function
    ELSE IF valid_choice is 3:
        CALL delete_task function
    ELSE IF valid_choice is 4:
        CALL save_tasks_to_file function
        PRINT "Tasks saved. Goodbye!"
        SET running = False
```

## Function: get_menu_choice()
```
FUNCTION get_menu_choice():
    WHILE True:
        TRY:
            PROMPT user for menu choice (1-4)
            GET choice_input from user input
            CONVERT choice_input to integer

            IF choice is between 1 and 4 (inclusive):
                RETURN choice
            ELSE:
                PRINT "Please enter a number between 1 and 4"
                CONTINUE to next iteration

        CATCH ValueError (if input is not a number):
            PRINT "Please enter a valid number"
            CONTINUE to next iteration
```
```
FUNCTION add_task():
    PROMPT user for task description
    GET task_description from user input

    IF task_description is not empty:
        CREATE new_task dictionary:
            id = task_counter
            description = task_description

        ADD new_task to tasks list
        INCREMENT task_counter by 1
        PRINT "Task added successfully!"
    ELSE:
        PRINT "Task description cannot be empty"
```

## Function: view_tasks()
```
FUNCTION view_tasks():
    IF tasks list is empty:
        PRINT "No tasks available"
        RETURN

    PRINT "Your Tasks:"
    FOR each task in tasks:
        PRINT task id and task description
        FORMAT: "ID: [id] - [description]"
```

## Function: delete_task()
```
FUNCTION delete_task():
    IF tasks list is empty:
        PRINT "No tasks to delete"
        RETURN

    CALL view_tasks to show current tasks

    GET task_id from get_valid_task_id function

    IF task_id is None:
        RETURN  # User cancelled or invalid input

    FOR each task in tasks:
        IF task id equals task_id:
            REMOVE task from tasks list
            PRINT "Task deleted successfully!"
            RETURN

    PRINT "Task with that ID not found"
```

## Helper Function: get_valid_task_id()
```
FUNCTION get_valid_task_id():
    WHILE True:
        TRY:
            PROMPT user for task ID to delete
            GET id_input from user input

            IF id_input is empty or user types "cancel":
                RETURN None

            CONVERT id_input to integer

            IF id is positive (greater than 0):
                RETURN id
            ELSE:
                PRINT "Task ID must be a positive number"
                CONTINUE to next iteration

        CATCH ValueError (if input is not a number):
            PRINT "Please enter a valid task ID number"
            CONTINUE to next iteration
```

## Error Handling Notes
- **Input Validation**: All numeric inputs are validated using try/except blocks
- **Menu Choice Validation**: Only accepts integers 1-4, keeps prompting until valid
- **Task ID Validation**: Ensures positive integers for task deletion
- **Graceful Loops**: Invalid input doesn't crash the program, just re-prompts
- **User-Friendly Messages**: Clear error messages guide the user
- **Exit Control**: Program only exits when user explicitly chooses option 4

## Key Improvements
- **Robust Input Handling**: Uses dedicated functions for validating different types of input
- **No Program Crashes**: Try/except blocks catch all input conversion errors
- **Persistent Menu**: Main loop continues running until explicit exit choice
- **Better User Experience**: Clear prompts and helpful error messages

## Sample Task Dictionary Structure
```
task = {
    "id": 1,
    "description": "Buy groceries"
}
```
````

# Section 1.2 - Second Refinement Prompt  


**Refinement Prompt 2:**  
Please refine the pseudocode to support loading tasks from a file at startup and saving them back to the file when exiting the program.

- Use open() to read/write from a .txt or .json file.

- Ensure task IDs are maintained between sessions.

- Display an error message if the file doesn’t exist and start with an empty task list.

- Use a while loop to create a persistent text-based dashboard that keeps prompting the user for actions (add, view, delete, exit) until they choose to quit.

````python
# Task Manager CLI - Pseudocode

## Data Structure
```
tasks = []  # List to store all tasks in memory
task_counter = 1  # To assign unique IDs to tasks
FILENAME = "tasks.json"  # File to store tasks persistently
```

## Main Program Flow
```
CALL load_tasks_from_file function

PRINT "=== Task Manager Dashboard ==="
SET running = True

WHILE running is True:
    PRINT "\n--- MENU ---"
    DISPLAY menu options:
        1. Add Task
        2. View All Tasks
        3. Delete Task
        4. Exit

    GET valid_choice from get_menu_choice function

    IF valid_choice is 1:
        CALL add_task function
    ELSE IF valid_choice is 2:
        CALL view_tasks function
    ELSE IF valid_choice is 3:
        CALL delete_task function
    ELSE IF valid_choice is 4:
        CALL save_tasks_to_file function
        PRINT "Tasks saved. Goodbye!"
        SET running = False
```

## Function: load_tasks_from_file()
```
FUNCTION load_tasks_from_file():
    GLOBAL tasks, task_counter

    TRY:
        OPEN FILENAME for reading
        READ file content
        PARSE JSON content into tasks list
        CLOSE file

        IF tasks list is not empty:
            FIND highest task ID in tasks
            SET task_counter = highest_id + 1
            PRINT "Loaded [number] tasks from file"
        ELSE:
            PRINT "No existing tasks found, starting fresh"

    CATCH FileNotFoundError:
        PRINT "No existing task file found, starting with empty task list"
        SET tasks = []
        SET task_counter = 1

    CATCH JSON decode error:
        PRINT "Error reading task file, starting with empty task list"
        SET tasks = []
        SET task_counter = 1

    CATCH any other error:
        PRINT "Error loading tasks, starting with empty task list"
        SET tasks = []
        SET task_counter = 1
```

## Function: save_tasks_to_file()
```
FUNCTION save_tasks_to_file():
    TRY:
        OPEN FILENAME for writing
        CONVERT tasks list to JSON format
        WRITE JSON content to file
        CLOSE file
        PRINT "Tasks saved successfully"

    CATCH any error:
        PRINT "Error saving tasks to file"
```
```
FUNCTION get_menu_choice():
    WHILE True:
        TRY:
            PROMPT user for menu choice (1-4)
            GET choice_input from user input
            CONVERT choice_input to integer

            IF choice is between 1 and 4 (inclusive):
                RETURN choice
            ELSE:
                PRINT "Please enter a number between 1 and 4"
                CONTINUE to next iteration

        CATCH ValueError (if input is not a number):
            PRINT "Please enter a valid number"
            CONTINUE to next iteration
```
```
FUNCTION add_task():
    PROMPT user for task description
    GET task_description from user input

    IF task_description is not empty:
        CREATE new_task dictionary:
            id = task_counter
            description = task_description

        ADD new_task to tasks list
        INCREMENT task_counter by 1
        PRINT "Task added successfully!"
    ELSE:
        PRINT "Task description cannot be empty"
```

## Function: view_tasks()
```
FUNCTION view_tasks():
    IF tasks list is empty:
        PRINT "No tasks available"
        RETURN

    PRINT "Your Tasks:"
    FOR each task in tasks:
        PRINT task id and task description
        FORMAT: "ID: [id] - [description]"
```

## Function: delete_task()
```
FUNCTION delete_task():
    IF tasks list is empty:
        PRINT "No tasks to delete"
        RETURN

    CALL view_tasks to show current tasks

    GET task_id from get_valid_task_id function

    IF task_id is None:
        RETURN  # User cancelled or invalid input

    FOR each task in tasks:
        IF task id equals task_id:
            REMOVE task from tasks list
            PRINT "Task deleted successfully!"

            # Optional: Auto-save after deleting
            CALL save_tasks_to_file function
            RETURN

    PRINT "Task with that ID not found"
```

## Helper Function: get_valid_task_id()
```
FUNCTION get_valid_task_id():
    WHILE True:
        TRY:
            PROMPT user for task ID to delete
            GET id_input from user input

            IF id_input is empty or user types "cancel":
                RETURN None

            CONVERT id_input to integer

            IF id is positive (greater than 0):
                RETURN id
            ELSE:
                PRINT "Task ID must be a positive number"
                CONTINUE to next iteration

        CATCH ValueError (if input is not a number):
            PRINT "Please enter a valid task ID number"
            CONTINUE to next iteration
```

## Error Handling Notes
- **File Operations**: All file read/write operations use try/except blocks
- **Missing File Handling**: Program gracefully starts with empty list if file doesn't exist
- **JSON Parsing**: Handles corrupted or invalid JSON files gracefully
- **Input Validation**: All numeric inputs are validated using try/except blocks
- **Menu Choice Validation**: Only accepts integers 1-4, keeps prompting until valid
- **Task ID Validation**: Ensures positive integers for task deletion
- **Graceful Loops**: Invalid input doesn't crash the program, just re-prompts
- **User-Friendly Messages**: Clear error messages guide the user
- **Exit Control**: Program only exits when user explicitly chooses option 4
- **Persistent Dashboard**: Menu keeps running in a loop until user chooses to quit

## Key Improvements
- **File Persistence**: Tasks are loaded at startup and saved on exit
- **ID Continuity**: Task counter maintains highest ID + 1 across sessions
- **Robust File Handling**: Handles missing files, corrupted JSON, and file errors
- **Dashboard Interface**: Clear menu structure with persistent loop
- **Auto-Save Options**: Can save after each add/delete operation
- **Error Recovery**: All file operations have fallback behaviors

## File Structure (JSON format)
```
[
    {
        "id": 1,
        "description": "Buy groceries"
    },
    {
        "id": 2,
        "description": "Walk the dog"
    }
]
```

## Sample Task Dictionary Structure
```
task = {
    "id": 1,
    "description": "Buy groceries"
}
```
````
